# Transformação e limpeza de dados

```python
# Remover colunas com 100% de valores nulos
df.dropna(axis=1, how='all', inplace=True)
```

```python

# Converter a coluna 'CL_FHL' para o tipo inteiro (int8)
df['CL_FHL'] = df['CL_FHL'].astype('int8')
```

```python

# Remover linhas duplicadas
df.drop_duplicates(inplace=True)
```

```python

# Verificar a distribuição da variável
print(f"Distribuição da variável 'CL_GENERO':\n{df['CL_GENERO'].value_counts(dropna=False)}\n")
print(f"Distribuição da variável 'CL_EC':\n{df['CL_EC'].value_counts(dropna=False)}\n")
print(f"Distribuição da variável 'CL_SEG':\n{df['CL_SEG'].value_counts(dropna=False)}\n")
print(f"Distribuição da variável 'PR_CAT':\n{df['PR_CAT'].value_counts(dropna=False)}\n")
```

```python
# Porcentagem de vendas por gênero em porcentagem
by_gender = df.groupby("CL_GENERO")["CO_ID"].nunique()
percentage = ((by_gender / by_gender.sum()) * 100).round(2)
print("Vendas por gênero (%):", percentage)
```

```python
# Porcentagem de vendas por categoria
category_percentage = (df["PR_CAT"].value_counts(normalize=True) * 100).round(2)
print("Vendas por categoria (%):", category_percentage)
```

```python
# Porcentagem das categorias por segmento social
all_segments = (df.groupby("CL_SEG")["PR_CAT"].value_counts(normalize=True) * 100).round(2)

for segment in df["CL_SEG"].unique():
    print(
        f"Vendas por categoria na Classe {segment} (%):\n{all_segments.loc[segment]}\n"
    )
```

```python
# Mostra a quantidade total de produtos vendidos por ano
for year, quantity in df.groupby(df["DATA"].dt.year).size().items():
    print(f"Ano {year}: Total de produtos vendidos: {quantity} unidades.")
```

```python
# Mostra os 3 produtod mais vendidos por estado civil


print(f"Produtos mais vendidos por estado civil\n")

for status in df["CL_EC"].unique():

    df_filtrado = df[df["CL_EC"] == status]

    top_3 = df_filtrado["PR_NOME"].value_counts().head(3)

    print(f">{status}")
    for product, quantity in top_3.items():
        print(f"- {product}: {quantity} unidades")
    print("\n")
```
